# Models Integration with OpenAI, Google-Gemini and GROQ

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")
model
```

```python
## invoke the model:
response = model.invoke("What is the capital of France?")
response 
```

```python
response.content
```

### Google Gemini Model Integration


```python
import os 
from langchain.chat_models import init_chat_model
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content

```

### GROQ Model Integration


```python
import os 
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen3-32b")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
##Use chatgroq:
import os
from langchain_groq import ChatGroq
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response.content
```

### Streaming


```python
for chunk in model.stream("write 200 words paragraph about the history of AI."):
    print(chunk.text, end = "|", flush = True)
```

### Batch

```python
responses = model.batch([
    "Why do parraots have colorful feathers?",
    "How do airplanes fly?",
    "what is quantum computing?"
])

for response in responses:
    print(response.content)
```

```python
model.batch([
    "Why do parraots have colorful feathers?",
    "How do airplanes fly?",
    "what is quantum computing?"],
    config = {
        "max_concurrency" : 5, #limit to 5 concurrent requests
    }
)
```
