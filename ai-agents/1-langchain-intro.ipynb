```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
import langchain
langchain.__version__
```

```python
from langchain.agents import create_agent

def get_weather(city:str) -> str:
    """Get the weather for a city"""
    return f"The weather in the {city} is cloudy"

agent = create_agent(
    model = "gpt-5",
    tools=[get_weather],
    system_prompt= "You are a helpful assistant",
)

agent
```

```python
### run the agent:

response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather in Bengaluru?"}]})

```
